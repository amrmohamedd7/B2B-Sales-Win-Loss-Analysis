```python
import pandas as pd
 
df = pd.read_csv("Salvirt_B2B_ML_dataset_HF.csv", sep=";")
print("Shape:", df.shape)
df.head()
```

```python
print(df["Status"].value_counts())
df["won"] = (df["Status"] == "Won").astype(int)
print(f"\nOverall win rate: {100 * df['won'].mean():.1f}%")
```

```python
for col in df.columns:
    n_unknown = (df[col] == "Unknown").sum()
    if n_unknown > 0:
        print(f"{col}: {n_unknown} rows ({100 * n_unknown / len(df):.1f}%) marked Unknown")
```

```python
n_dupes = df.duplicated(subset=df.columns.drop("won")).sum()
print(f"\nExact duplicate rows: {n_dupes} ({100 * n_dupes / len(df):.1f}%)")
```

```python
print("\nUnique products:", df["Product"].nunique())
print("Unique sellers:", df["Seller"].nunique())
print(df["Seller"].value_counts().head(10))
```

```python
df.to_csv("b2b_sales_clean.csv", index=False)
print("\nSaved b2b_sales_clean.csv")
```
